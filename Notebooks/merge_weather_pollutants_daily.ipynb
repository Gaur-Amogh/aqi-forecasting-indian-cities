```python
import pandas as pd
```

```python
weather = pd.read_csv("../Datasets/Delhi/Delhi-weather.csv")
pollutants = pd.read_csv("../Datasets/Delhi/Delhi-pollutants.csv")
```

```python
print(weather.head())
print(pollutants.head())
```

```python
weather = weather[
    [
        "time",
        "temperature_2m (°C)",
        "relative_humidity_2m (%)",
        "wind_speed_10m (km/h)"
    ]
]
```

```python
pollutants = pollutants[
    [
        "Timestamp",
        "PM2.5 (µg/m³)",
        "PM10 (µg/m³)",
        "NO2 (µg/m³)",
        "SO2 (µg/m³)",
        "CO (mg/m³)",
        "Ozone (µg/m³)"
    ]
]
```

```python
weather["date"] = pd.to_datetime(weather["time"]).dt.date
pollutants["date"] = pd.to_datetime(pollutants["Timestamp"]).dt.date
```

```python
weather_daily = (
    weather
    .groupby("date", as_index=False)
    .mean(numeric_only=True)
)

pollutants_daily = (
    pollutants
    .groupby("date", as_index=False)
    .mean(numeric_only=True)
)
```

```python
delhi_merged = pd.merge(
    pollutants_daily,
    weather_daily,
    on="date",
    how="inner"
)
```

```python
print(delhi_merged.head())
print(delhi_merged.isnull().sum())
print(delhi_merged.shape)
```

```python
import os

os.makedirs("../outputs", exist_ok=True)
```

```python
delhi_merged.to_csv("../outputs/delhi_merged_daily.csv", index=False)
```
